**Why:** drop columns with >50% missing — too sparse to impute reliably.

```python
# Drop high precentage nulls columns
limit = 50.0
columns_to_drop = missing_percentages[missing_percentages > limit].index

df_cleaned = df.drop(columns=columns_to_drop)

print(f"Original number of columns: {df.shape[1]}")
print(f"Number of columns dropped: {len(columns_to_drop)}")
print(f"Remaining columns: {df_cleaned.shape[1]}")
```

**Why:** split remaining columns by type so imputation strategy matches data type.

```python
from sklearn.impute import SimpleImputer

# Group columns by data type
num_cols = df_cleaned.select_dtypes(include=['float64', 'int64']).columns
cat_cols = df_cleaned.select_dtypes(include=['object']).columns

# Remove Target Variable 
num_cols = num_cols.drop('hospital_death')
    
print(f"Found {len(num_cols)} numerical columns and {len(cat_cols)} categorical columns.")
```

## Feature Engineering

**Why:** add clinically motivated risk flags (hypotension, tachycardia, hypoxia) based on known danger thresholds.

```python
# Create Binary Risk Flags based on clinical thresholds
# Flag 1 if minimum systolic BP drops dangerously low (Hypotension)
df_encoded['risk_hypotension'] = (df_encoded['d1_sysbp_min'] < 90).astype(int)

# Flag 1 if max heart rate goes dangerously high (Tachycardia)
df_encoded['risk_tachycardia'] = (df_encoded['d1_heartrate_max'] > 120).astype(int)

# Flag 1 if minimum oxygen drops dangerously low (Hypoxia)
df_encoded['risk_hypoxia'] = (df_encoded['d1_spo2_min'] < 90).astype(int)

print("Engineered 3 new 'Clinical Risk Flag' features successfully.")
```

**Why:** engineer clinical ratios (shock index, BUN/creatinine, SpO2/resp rate) that are known risk indicators but aren't in the raw data.

```python
# Shock Index (Max HR / Min Systolic BP)
# We add 1e-5 to prevent division by zero errors
df_encoded['shock_index'] = df_encoded['d1_heartrate_max'] / (df_encoded['d1_sysbp_min'] + 1e-5)

# BUN to Creatinine Ratio (Max BUN / Max Creatinine)
# Checks for acute kidney failure
if 'd1_bun_max' in df_encoded.columns and 'd1_creatinine_max' in df_encoded.columns:
    df_encoded['bun_creatinine_ratio'] = df_encoded['d1_bun_max'] / (df_encoded['d1_creatinine_max'] + 1e-5)

# SpO2 to Respiratory Rate Ratio (Approximation for respiratory distress)
df_encoded['spo2_resprate_ratio'] = df_encoded['d1_spo2_min'] / (df_encoded['h1_resprate_max'] + 1e-5)

print("Clinical ratios added successfully.")
```

**Why:** aggregate GCS into one score and add vital-sign ranges/volatility — a large range can signal instability better than max or min alone.

```python
# Total Glasgow Coma Scale (GCS)
gcs_cols = ['gcs_motor_apache', 'gcs_eyes_apache', 'gcs_verbal_apache']
if all(col in df_encoded.columns for col in gcs_cols):
    df_encoded['gcs_total_score'] = df_encoded[gcs_cols].sum(axis=1)

# Calculate Ranges for core vitals (Max - Min)
vitals = ['d1_heartrate', 'd1_sysbp', 'd1_temp', 'd1_spo2']
for v in vitals:
    if f'{v}_max' in df_encoded.columns and f'{v}_min' in df_encoded.columns:
        df_encoded[f'{v}_range'] = df_encoded[f'{v}_max'] - df_encoded[f'{v}_min']

# Extreme Volatility Flag (Count how many vitals had a massive swing)
# E.g., HR range > 40, SysBP range > 40, Temp range > 2 degrees
df_encoded['high_volatility_count'] = (
    (df_encoded.get('d1_heartrate_range', 0) > 40).astype(int) +
    (df_encoded.get('d1_sysbp_range', 0) > 40).astype(int) +
    (df_encoded.get('d1_temp_range', 0) > 2.0).astype(int)
)

print("Aggregate scores and volatility tracking added successfully.")
```

**Why:** log-transform right-skewed lab values so their distribution is more model-friendly.

```python
# Features that are typically highly right-skewed in medical data
skewed_features = ['d1_glucose_max', 'd1_bun_max', 'd1_creatinine_max', 'd1_bilirubin_max']

for col in skewed_features:
    if col in df_encoded.columns:
        # We use log1p (log(1 + x)) to handle any zeros safely
        df_encoded[f'{col}_log'] = np.log1p(df_encoded[col])
        
print("Log transformations completed.")
```

**Why:** split into train/test **before** feature selection to prevent leakage — test set stays untouched until final evaluation.

```python
from sklearn.model_selection import train_test_split

# Drop non-predictive IDs
X_full = df_encoded.drop(columns=['hospital_death', 'encounter_id', 'patient_id'], errors='ignore')
y_full = df_encoded['hospital_death']

# Strict split BEFORE feature selection
X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.2, random_state=42, stratify=y_full
)

print(f"Training Data Shape: {X_train.shape}")
print(f"Testing Data Shape (Locked away): {X_test.shape}")
```

**Why:** use Mutual Information to cheaply narrow ~150 features down to the 80 most informative, before the slower RFE step.

```python
from sklearn.feature_selection import mutual_info_classif

# Calculate MI scores
mi_scores = mutual_info_classif(X_train.fillna(0), y_train, random_state=42)

# Create a sorted Series
mi_series = pd.Series(mi_scores, index=X_train.columns).sort_values(ascending=False)

# Filter down to the Top 80 features for the next stage
top_80_features = mi_series.head(80).index.tolist()
X_train_mi = X_train[top_80_features]
```

```python
# Plot the Top 20 to see what the math found
plt.figure(figsize=(12, 8))
sns.barplot(x=mi_series.head(20).values, y=mi_series.head(20).index, palette='magma')
plt.title("Top 20 Features by Mutual Information Score", fontsize=14)
plt.xlabel("Information Gain")
plt.ylabel("Feature Name")
plt.show()
```

**Why:** run RFE on the MI-reduced set to select the final 40 features — two-stage selection keeps RFE fast.

```python
from sklearn.feature_selection import RFE

# We use a fast, shallow Random Forest as the judge for RFE to save time
rfe_judge = RandomForestClassifier(n_estimators=30, max_depth=5, random_state=42, n_jobs=-1)

rfe = RFE(estimator=rfe_judge, n_features_to_select=40, step=2)
rfe.fit(X_train_mi.fillna(0), y_train)

final_40_features = X_train_mi.columns[rfe.support_].tolist()

print("\n--- RFE COMPLETE ---")
print(f"Successfully isolated the optimal {len(final_40_features)} clinical features.")

X_train_final = X_train[final_40_features]
X_test_final = X_test[final_40_features]

print(f"Final X_train shape: {X_train_final.shape}")
```

**Why:** switch from SMOTE to XGBoost's built-in `scale_pos_weight` — avoids synthetic patient records and gives better-calibrated probabilities for a tree model.

## Handling Imbalance: scale_pos_weight

SMOTE was dropped in favor of XGBoost's built-in `scale_pos_weight`, since it avoids synthetic patient records and tends to produce better-calibrated probabilities for tree models.

**Why:** compute the class ratio XGBoost needs to internally upweight the rare mortality class.

```python
neg, pos = y_train.value_counts()[0], y_train.value_counts()[1]
scale_pos_weight = neg / pos
print(f"scale_pos_weight = {scale_pos_weight:.2f}")
```

**Why:** train XGBoost on the real (unbalanced) training data, weighted rather than resampled.

```python
from xgboost import XGBClassifier

xgb_weighted = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    eval_metric='logloss',
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)
xgb_weighted.fit(X_train_final.fillna(0), y_train)
```

```python
X_test_final_clean = X_test_final.fillna(0)
y_pred_w = xgb_weighted.predict(X_test_final_clean)
y_proba_w = xgb_weighted.predict_proba(X_test_final_clean)[:, 1]
```

```python
from sklearn.metrics import classification_report, roc_auc_score,average_precision_score, confusion_matrix, brier_score_loss

print("=== scale_pos_weight model (default 0.5 threshold) ===\n")
print(classification_report(y_test, y_pred_w, target_names=['Survived', 'Died']))
print(f"AUROC: {roc_auc_score(y_test, y_proba_w):.4f}")
print(f"AUPRC: {average_precision_score(y_test, y_proba_w):.4f}")
print(f"Brier Score: {brier_score_loss(y_test, y_proba_w):.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_w))
```

**Why:** the default 0.5 threshold under-predicts mortality (low recall). Sweep thresholds to see the precision/recall trade-off.

```python
from sklearn.metrics import precision_recall_curve

precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba_w)

pr_table = pd.DataFrame({
    'threshold': thresholds,
    'precision': precisions[:-1],
    'recall': recalls[:-1]
})
pr_table['f1'] = 2 * pr_table['precision'] * pr_table['recall'] / (pr_table['precision'] + pr_table['recall'])

# Show thresholds where recall lands in a clinically useful range
print(pr_table[(pr_table['recall'] >= 0.6) & (pr_table['recall'] <= 0.8)].sort_values('threshold').head(15))
```

**Why:** compare precision/recall/F1 across candidate thresholds side-by-side.

```python
for t in [0.25, 0.30, 0.35, 0.40, 0.45, 0.50]:
    y_pred_t = (y_proba_w >= t).astype(int)
    from sklearn.metrics import precision_score, recall_score, f1_score
    p = precision_score(y_test, y_pred_t)
    r = recall_score(y_test, y_pred_t)
    f1 = f1_score(y_test, y_pred_t)
    print(f"threshold={t:.2f}  precision={p:.3f}  recall={r:.3f}  f1={f1:.3f}")
```

**Why:** lock in the chosen threshold (0.40) — balances catching more at-risk patients against an acceptable false-alarm rate.

```python
# Pick a threshold from the table above (edit this after inspecting the printout)
chosen_threshold = 0.40

y_pred_tuned = (y_proba_w >= chosen_threshold).astype(int)

print(f"=== scale_pos_weight model (threshold = {chosen_threshold}) ===\n")
print(classification_report(y_test, y_pred_tuned, target_names=['Survived', 'Died']))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_tuned))
```
